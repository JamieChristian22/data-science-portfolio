## Baseline and challengers

Compare interpretable and nonlinear candidates using PR-AUC and ROC-AUC.

```python
import sys
sys.path.append('..')
from src.credit_risk_pipeline import load_data, make_candidates
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, average_precision_score

df=load_data('../data/credit_risk.csv'); X=df.drop(columns='defaulted'); y=df.defaulted
Xtr,Xv,ytr,yv=train_test_split(X,y,test_size=.25,random_state=42,stratify=y)
for name,model in make_candidates(X).items():
    model.fit(Xtr.drop(columns=['application_id']),ytr)
    p=model.predict_proba(Xv.drop(columns=['application_id']))[:,1]
    print(name, round(roc_auc_score(yv,p),3), round(average_precision_score(yv,p),3))
```

## Why PR-AUC?

Defaults are the minority class, so PR-AUC provides useful information beyond accuracy.
