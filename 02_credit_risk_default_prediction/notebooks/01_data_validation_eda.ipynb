# 01 — Data Validation & EDA

## Objective

Validate schema, inspect class balance, and understand the synthetic credit-risk population.

```python
from pathlib import Path
import pandas as pd
ROOT=Path('..')
df=pd.read_csv(ROOT/'data/credit_risk.csv')
df.head()
```

## Quality checks

Check duplicates, missingness, target rate, and numeric summaries.

```python
print(df.shape)
print('duplicates:', df.duplicated().sum())
print(df.isna().sum())
print('default rate:', df.defaulted.mean())
df.describe(include='all').T
```

## Risk relationships

Compare default rates across important features. Association is not causation.

```python
pd.qcut(df.credit_score,5,duplicates='drop').to_frame('score_band').join(df.defaulted).groupby('score_band', observed=False).defaulted.mean()
```
