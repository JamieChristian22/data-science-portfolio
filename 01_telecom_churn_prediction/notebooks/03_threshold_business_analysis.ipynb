# 03 — Threshold & Business Analysis
Optimize the classification threshold for churn outreach rather than accepting the default 0.50 cutoff.

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split
from src.modeling import FEATURES, candidate_models, choose_threshold
ROOT=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
df=pd.read_csv(ROOT/'data/telecom_churn.csv'); X=df[FEATURES]; y=df.churned
Xtr,Xv,ytr,yv=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
m=candidate_models()['Logistic Regression']; m.fit(Xtr,ytr); p=m.predict_proba(Xv)[:,1]
t,table=choose_threshold(yv,p,min_precision=.28); print('Selected threshold',t); table.sort_values('f2',ascending=False).head(10)
```

F2 gives recall more weight than precision, which fits a retention use case where missing a true churner can be more costly than contacting some customers who would have stayed.
