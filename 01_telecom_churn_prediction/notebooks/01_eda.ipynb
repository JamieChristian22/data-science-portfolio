# 01 — Exploratory Data Analysis
Business-oriented inspection of churn prevalence, data quality, and primary risk patterns.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
df=pd.read_csv(ROOT/'data/telecom_churn.csv')
df.head()
```

```python
df.info(); print('Churn rate:', df.churned.mean()); df.describe(include='all').T
```

```python
df.groupby('contract_type').churned.mean().sort_values().plot.bar(title='Churn Rate by Contract Type'); plt.ylabel('Churn rate'); plt.show()
```

```python
df.groupby(pd.qcut(df['satisfaction_score_1to5'],4)).churned.mean().plot.bar(title='Churn by Satisfaction Quartile'); plt.ylabel('Churn rate'); plt.show()
```

## EDA takeaway
The analysis focuses on actionable behavioral and service variables rather than customer identifiers. The modeling pipeline excludes `customer_id` to prevent identifier leakage.
