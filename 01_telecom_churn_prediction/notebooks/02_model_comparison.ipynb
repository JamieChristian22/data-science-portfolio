# 02 — Model Comparison
Compare an interpretable baseline against nonlinear ensemble models using cross-validation and validation-set metrics.

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split
from src.modeling import FEATURES, candidate_models, classification_metrics, cv_summary
ROOT=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
df=pd.read_csv(ROOT/'data/telecom_churn.csv'); X=df[FEATURES]; y=df.churned
Xtr,Xv,ytr,yv=train_test_split(X,y,test_size=.25,stratify=y,random_state=42)
```

```python
rows=[]
for name,m in candidate_models().items():
    m.fit(Xtr,ytr); p=m.predict_proba(Xv)[:,1]
    rows.append({'model':name, **classification_metrics(yv,p,.5), **cv_summary(m,Xtr,ytr)})
pd.DataFrame(rows).sort_values('pr_auc',ascending=False)
```
